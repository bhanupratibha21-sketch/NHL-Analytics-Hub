```python
import time
import requests

# Create an empty list to store players from all NHL teams
all_players_data = []

# Fetch the current roster for every team
for team_abbrev in teams_df["team_abbrev"]:

    # Build the roster API URL for the current team
    roster_url = f"https://api-web.nhle.com/v1/roster/{team_abbrev}/current"

    # Send the API request
    response = requests.get(roster_url)

    # Check whether the API request was successful
    if response.status_code == 200:

        # Convert the API response into a Python dictionary
        roster = response.json()

        # Combine all player categories
        players = (
            roster.get("forwards", [])
            + roster.get("defensemen", [])
            + roster.get("goalies", [])
        )

        # Extract required player information
        for player in players:
            all_players_data.append({
                "team_abbrev": team_abbrev,
                "player_id": player["id"],
                "first_name": player["firstName"]["default"],
                "last_name": player["lastName"]["default"],
                "position": player["positionCode"],
                "jersey_number": player.get("sweaterNumber"),
                "birth_date": player.get("birthDate"),
                "birth_country": player.get("birthCountry"),
                "height_cm": player.get("heightInCentimeters"),
                "weight_kg": player.get("weightInKilograms"),
                "shoots_catches": player.get("shootsCatches"),
                "headshot_url": player.get("headshot")
            })

        print(f"{team_abbrev}: {len(players)} players fetched")

    else:
        print(f"{team_abbrev}: API request failed - {response.status_code}")

    # Add a small delay between API requests
    time.sleep(0.5)

# Convert all collected player records into a DataFrame
players_df = pd.DataFrame(all_players_data)

# Display the total number of collected players
print("\nTotal players collected:", len(players_df))
```

```python
# Create an empty list to store game records from all NHL teams
all_games_data = []

# Fetch the season schedule for every NHL team
for team_abbrev in teams_df["team_abbrev"]:

    # Build the schedule API URL
    schedule_url = (
        f"https://api-web.nhle.com/v1/club-schedule-season/"
        f"{team_abbrev}/20252026"
    )

    # Send the API request
    response = requests.get(schedule_url)

    # Process the response only when the request is successful
    if response.status_code == 200:

        # Convert the JSON response into a Python dictionary
        schedule = response.json()

        # Extract every game from the team's schedule
        for game in schedule.get("games", []):

            all_games_data.append({
                "game_id": game["id"],
                "season": str(game["season"]),
                "game_type": game["gameType"],
                "game_date": game["gameDate"],
                "home_team_abbrev": game["homeTeam"]["abbrev"],
                "away_team_abbrev": game["awayTeam"]["abbrev"],
                "home_score": game["homeTeam"].get("score"),
                "away_score": game["awayTeam"].get("score"),
                "game_state": game["gameState"],
                "venue_name": game["venue"].get("default")
            })

        print(f"{team_abbrev}: {len(schedule.get('games', []))} games fetched")

    else:
        print(f"{team_abbrev}: API request failed - {response.status_code}")

    time.sleep(0.5)

# Convert all collected game records into a DataFrame
games_df = pd.DataFrame(all_games_data)

# Display the total number of collected game records
print("\nTotal game records before deduplication:", len(games_df))
```
